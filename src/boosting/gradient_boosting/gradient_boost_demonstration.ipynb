```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
```

```python
X1 = np.linspace(-10, 10, 100)
X2= 2* X1
Y= X1**2 + X2

df= pd.DataFrame(np.c_[X1, X2, Y], columns= ['x1', 'x2', 'y'])
```

```python
#initial data plot
x, y = df['x1'], df['y'] 
plt.scatter(x, y)
```

```python
# initial model (f0)
f0= np.mean(y)

plt.scatter(x, y)
plt.axhline(f0, x.min(), x.max(), linewidth= 2.0, color= 'red')
```

```python
res_f0 = y- f0
dt= DecisionTreeRegressor(max_leaf_nodes= 8)
dt.fit(df[['x1', 'x2']], res_f0)
```

```python
f1= f0 + dt.predict(df[['x1', 'x2']])
plt.scatter(x, y)
plt.plot(x ,f1, color= 'red', linewidth= 2.0)
```

```python
#one more iteration
res_f1= y - f1
dt2= DecisionTreeRegressor(max_leaf_nodes= 8)
dt2.fit(df[['x1', 'x2']], res_f1)
```

```python
f2= f1 + dt2.predict(df[['x1', 'x2']])
plt.scatter(x, y)
plt.plot(x ,f2, color= 'red', linewidth= 2.0)
```

```python
#one_last_iteration
res_f2= y - f2
dt3= DecisionTreeRegressor(max_leaf_nodes= 8)
dt3.fit(df[['x1', 'x2']], res_f2)
```

```python
f3= f2 + dt3.predict(df[['x1', 'x2']])
plt.scatter(x, y)
plt.plot(x ,f3, color= 'red', linewidth= 2.0)
```
